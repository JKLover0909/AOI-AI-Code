```python
import cv2
import numpy as np
import os

def convert_gray_to_white(image_path, output_path):
    # Đọc ảnh
    image = cv2.imread(image_path, cv2.IMREAD_COLOR)
    if image is None:
        print(f"Không thể đọc ảnh: {image_path}")
        return

    # Chuyển ảnh sang không gian màu HSV
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

    # Định nghĩa ngưỡng màu xám (giá trị có thể điều chỉnh tùy thuộc vào ảnh)
    lower_gray = np.array([0, 0, 50])  # Ngưỡng dưới (H, S, V)
    upper_gray = np.array([180, 50, 200])  # Ngưỡng trên (H, S, V)

    # Tạo mặt nạ cho vùng màu xám
    mask = cv2.inRange(hsv, lower_gray, upper_gray)

    # Thay đổi vùng màu xám thành màu trắng
    image[mask > 0] = [255, 255, 255]

    # Lưu ảnh đã xử lý
    cv2.imwrite(output_path, image)

def process_folder(input_folder, output_folder):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    for root, _, files in os.walk(input_folder):
        for file in files:
            if file.lower().endswith(('.bmp', '.jpg', '.png')):
                input_path = os.path.join(root, file)
                output_path = os.path.join(output_folder, file)
                convert_gray_to_white(input_path, output_path)

# Thay 'input_folder' bằng đường dẫn tới folder chứa ảnh gốc
input_folder = r'D:\Data\DataVRS\Data3'
# Thay 'output_folder' bằng đường dẫn tới folder lưu ảnh đã xử lý
output_folder = r'D:\Data\DataVRS\Dataafter'

process_folder(input_folder, output_folder)

print(f"Đã xử lý xong ảnh từ {input_folder} và lưu vào {output_folder}")
```
